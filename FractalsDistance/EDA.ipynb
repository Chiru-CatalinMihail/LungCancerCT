# Import libraries

```python
import os
import warnings
warnings.filterwarnings("ignore") # remove some scikit-image warnings


import torch
import numpy as np
import matplotlib.pyplot as plt
import random
import cv2
```

```python
crt_path = os.getcwd() + '/' + 'fractal map'
crt_path
```

```python
files = os.listdir(crt_path)
files = [crt_path + '/' + file for file in files]
```

```python
import torch
```

```python
import sys

for file in files:
    abc = torch.load(file)

    idx = 0

    print(abc.shape, torch.min(abc[idx]), torch.max(abc[idx]))

    fig, axes = plt.subplots(1, 4, figsize=(10, 5))

    axes[0].imshow(abc[0, :, :])
    axes[1].imshow(abc[1, :, :])
    axes[2].imshow(abc[0, :, :])
    axes[2].imshow(abc[2, :, :], alpha=0.5)
    axes[3].imshow(abc[3, :, :])

    plt.show()

    print(abc.shape)
    sys.exit(0)


```
